```python
import os
import zipfile
from pathlib import Path

ZIP_PATH = "/content/drive/MyDrive/Blink test.zip"
OUTPUT_DIR = "/content/drive/MyDrive/MRL_Combined"

# Create fresh output folders
os.makedirs(OUTPUT_DIR, exist_ok=True)

for cls in ["awake", "sleepy"]:
    os.makedirs(os.path.join(OUTPUT_DIR, cls), exist_ok=True)

counts = {"awake": 0, "sleepy": 0}
extensions = {".png", ".jpg", ".jpeg", ".bmp", ".webp"}

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    for member in z.infolist():
        if member.is_dir():
            continue

        parts = Path(member.filename).parts
        if len(parts) < 3:
            continue

        # Find images belonging to either class in any split
        cls = parts[-2].lower()

        if cls not in counts:
            continue

        if Path(member.filename).suffix.lower() not in extensions:
            continue

        counts[cls] += 1
        output_name = f"{counts[cls]:06d}_{Path(member.filename).name}"
        output_path = os.path.join(OUTPUT_DIR, cls, output_name)

        with z.open(member) as src, open(output_path, "wb") as dst:
            dst.write(src.read())

print("\nDataset combined successfully!")
print("Awake images:", counts["awake"])
print("Sleepy images:", counts["sleepy"])
print("Total images:", sum(counts.values()))
print("Saved in:", OUTPUT_DIR)
```

```python
import os
import random
import shutil
from pathlib import Path

SOURCE_DIR = "/content/drive/MyDrive/MRL_Combined"
BALANCED_DIR = "/content/drive/MyDrive/MRL_Balanced"

random.seed(42)

# Get image files
awake_files = [
    p for p in (Path(SOURCE_DIR) / "awake").iterdir()
    if p.is_file()
]

sleepy_files = [
    p for p in (Path(SOURCE_DIR) / "sleepy").iterdir()
    if p.is_file()
]

# Balance both classes to the smaller class size
target_count = min(len(awake_files), len(sleepy_files))

selected_awake = random.sample(awake_files, target_count)
selected_sleepy = random.sample(sleepy_files, target_count)

# Recreate balanced output folders
if os.path.exists(BALANCED_DIR):
    shutil.rmtree(BALANCED_DIR)

for cls, selected_files in [
    ("awake", selected_awake),
    ("sleepy", selected_sleepy)
]:
    output_folder = Path(BALANCED_DIR) / cls
    output_folder.mkdir(parents=True, exist_ok=True)

    for i, file_path in enumerate(selected_files):
        shutil.copy2(file_path, output_folder / f"{i:06d}_{file_path.name}")

print("Balanced dataset created!")
print("Awake:", len(selected_awake))
print("Sleepy:", len(selected_sleepy))
print("Total:", len(selected_awake) + len(selected_sleepy))
print("Location:", BALANCED_DIR)
```
